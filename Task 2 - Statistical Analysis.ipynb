```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Task 2: Statistical Analysis & Visualization")
print("Libraries imported successfully!")
```

```python
data = {
    "Student": ["Anu", "Bala", "Charu", "Deepa", "Ezhil",
                "Farah", "Hari", "Isha", "Jaya", "Kavi"],
    "Study_Hours": [2, 4, 3, 5, 6, 2, 7, 4, 5, 3],
    "Marks": [45, 60, 55, 75, 85, 50, 90, 65, 80, 58],
    "Attendance": [70, 80, 75, 90, 95, 72, 98, 85, 92, 78]
}

df = pd.DataFrame(data)

print("Student Dataset:")
display(df)

print("\nDataset Shape:", df.shape)
```

```python
print("Statistical Analysis")
print("--------------------")

print("\nMean:")
display(df[["Study_Hours", "Marks", "Attendance"]].mean())

print("\nMedian:")
display(df[["Study_Hours", "Marks", "Attendance"]].median())

print("\nStandard Deviation:")
display(df[["Study_Hours", "Marks", "Attendance"]].std())

print("\nMinimum Values:")
display(df[["Study_Hours", "Marks", "Attendance"]].min())

print("\nMaximum Values:")
display(df[["Study_Hours", "Marks", "Attendance"]].max())

print("\nComplete Summary:")
display(df.describe())
```

```python
# 1. Study Hours vs Marks
plt.figure(figsize=(7, 4))
plt.scatter(df["Study_Hours"], df["Marks"])
plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.grid(True)
plt.show()

# 2. Marks Distribution
plt.figure(figsize=(7, 4))
plt.hist(df["Marks"], bins=5, edgecolor="black")
plt.title("Distribution of Student Marks")
plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.show()

# 3. Correlation Heatmap
plt.figure(figsize=(7, 4))
sns.heatmap(
    df[["Study_Hours", "Marks", "Attendance"]].corr(),
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)
plt.title("Correlation Between Variables")
plt.show()
```

```python
print("KEY FINDINGS")
print("=" * 35)

print("Average Study Hours:", round(df["Study_Hours"].mean(), 2))
print("Average Marks:", round(df["Marks"].mean(), 2))
print("Average Attendance:", round(df["Attendance"].mean(), 2), "%")

correlation = df["Study_Hours"].corr(df["Marks"])
print("Study Hours and Marks Correlation:", round(correlation, 2))

if correlation > 0:
    print("Finding: Higher study hours are associated with higher marks.")
else:
    print("Finding: Study hours do not show a positive relationship with marks.")

print("\nAnalysis completed successfully!")
```

```python
df.to_csv("Student_Statistical_Data.csv", index=False)

print("Dataset saved successfully!")
```
