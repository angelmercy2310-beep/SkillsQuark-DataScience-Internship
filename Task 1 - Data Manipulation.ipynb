```python
import pandas as pd
import numpy as np

data = {
    "Customer_ID": [101, 102, 103, 104, 105, 106, 106, 108, 109, 110],
    "Name": ["Anu", "Bala", "Charu", "Deepa", "Ezhil",
             "Farah", "Farah", "Hari", "Isha", "Jaya"],
    "Age": [22, 25, np.nan, 28, 30, 24, 24, np.nan, 27, 32],
    "City": ["Chennai", "Madurai", "Chennai", "Trichy", np.nan,
             "Salem", "Salem", "Chennai", "Madurai", "Trichy"],
    "Purchase_Amount": [1500, 2500, 1800, np.nan, 3200,
                        2100, 2100, 1700, np.nan, 4000]
}

df = pd.DataFrame(data)

print("Original Dataset:")
display(df)

print("Dataset Shape:", df.shape)
print("Missing Values:")
print(df.isnull().sum())
```

```python
# Remove duplicate records
df = df.drop_duplicates()

print("Dataset after removing duplicates:")
display(df)

print("Updated Dataset Shape:", df.shape)
print("Duplicate records remaining:", df.duplicated().sum())
```

```python
# Fill missing Age with the median
df["Age"] = df["Age"].fillna(df["Age"].median())

# Fill missing City with the most frequent city
df["City"] = df["City"].fillna(df["City"].mode()[0])

# Fill missing Purchase_Amount with the median
df["Purchase_Amount"] = df["Purchase_Amount"].fillna(
    df["Purchase_Amount"].median()
)

print("Dataset after filling missing values:")
display(df)

print("\nMissing values remaining:")
print(df.isnull().sum())
```

```python
# Convert Age into integer format
df["Age"] = df["Age"].astype(int)

# Create a new column for customer age group
df["Age_Group"] = df["Age"].apply(
    lambda age: "Young" if age < 25
    else "Adult" if age <= 30
    else "Senior"
)

# Create a new column with a 10% discount
df["Discount_Amount"] = df["Purchase_Amount"] * 0.10

print("Dataset after column transformation:")
display(df)

print("Updated columns:")
print(df.columns.tolist())
```

```python
print("Final Dataset Information")
print("-------------------------")

print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate records:", df.duplicated().sum())

print("\nStatistical Summary:")
display(df.describe())
```

```python
# Save the cleaned dataset as a CSV file
df.to_csv("Cleaned_Customer_Data.csv", index=False)

print("Cleaned dataset saved successfully!")
print("Final dataset shape:", df.shape)

display(df.head())
```
